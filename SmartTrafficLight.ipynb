```python
import time

def traffic_light(night_mode=False, pedestrian_pressed=False):
    if night_mode:
        green_time = 10
        yellow_time = 5
        red_time = 10
    else:
        green_time = 15
        yellow_time = 10
        red_time = 15

    print("Green Light On")
    time.sleep(green_time)

    print("Yellow Light On")
    time.sleep(yellow_time)

    print("Red Light On")
    if pedestrian_pressed:
        print("WALK SIGNAL ON")
        time.sleep(3)
        print("WALK SIGNAL OFF")
    time.sleep(red_time)

while True:
    mode = input("Enter mode (day/night): ")
    pedestrian = input("Pedestrian pressed? (yes/no): ")

    traffic_light(
        night_mode = (mode == "night"),
        pedestrian_pressed = (pedestrian == "yes")
    )
```
